# VidTextSwap: replace text in videos and images, keeping the style

Change a **phone number, branch name, website or offer date** inside your videos and images.
The new text keeps the original size, colour, outline and boldness, AI rebuilds the background
behind the old text, and videos keep their sound.

**How to use this notebook (free, about 3 minutes):**
1. Click **Runtime → Run all** (or press the ▶ button on each cell, top to bottom).
2. Wait for the cell **"2. Start the app"** to print a link like `https://xxxx.gradio.live`.
3. Open that link: upload files, type what to change, and download the results.

This runs on Google's free machines, so nothing is installed on your computer.
Your files go only to this Colab session, which is deleted when you close it.
